```python
import networkx as nx
import pyrigi
from pyrigi import Graph
import networkx as nx
import matplotlib.pyplot as plt
from project_utils.rewards import NumNAC, NumPlane, NumSphere
```

```python
def plot_graphs(best_graphs, reward_func, caption):
    num_graphs = len(best_graphs)
    ncols = min(3, num_graphs)
    nrows = (num_graphs + 2) // 3
    
    fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(16, 5 * nrows), squeeze=False)
    ax_list = axes.flatten()

    for i, graph in enumerate(best_graphs):
        ax = ax_list[i]
        G = Graph.from_int(graph)
        n_vertices = G.number_of_nodes()
        title_text = f"Vertices: {n_vertices}"
        ax.set_title(title_text, fontsize=12, color="black", fontweight="bold")

        desc = f"{caption}: {reward_func(graph, n_vertices)}\n{graph}"
        ax.set_xlabel(desc, fontsize=10, style='italic', color='gray')

        pos = nx.spring_layout(G, seed=42) 
        nx.draw(G, pos, ax=ax)
        
        ax.set_axis_on()
        ax.tick_params(left=False, bottom=False, labelleft=False, labelbottom=False)
```

```python
best_graphs_plane = [7171569989696, 2984633745867526, 14279390632815629528, 191262807660326979451400,
                     1282672888403406155507302440, 20976294426015366843182805451832, 64952517330210831777993107755504388,
                     13953262603628863561574116134845685961796, 5821888507584727466754147148777408623257330338]
plot_graphs(best_graphs_plane, NumPlane(), "Plane#")
```
